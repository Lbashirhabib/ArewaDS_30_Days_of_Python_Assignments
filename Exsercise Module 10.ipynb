### Level 1 - Answers

**1. Difference between map, filter and reduce**

- `map` takes a function and a list, applies the function to every item, and gives back a new list of the same size.
- `filter` takes a function that returns True or False, and keeps only the items where it returns True. The result can be shorter than the original list.
- `reduce` takes a function with two inputs and keeps combining the items one by one until only a single value is left (for example a sum).

**2. Higher order function, closure and decorator**

- A *higher order function* is a function that takes another function as an input, or returns a function. `map`, `filter` and `reduce` are examples.
- A *closure* is an inner function that remembers the variables of the outer function, even after the outer function has finished running.
- A *decorator* is a function that takes another function and adds extra behaviour to it without changing its code. It is written with the `@` symbol above a function.

**3.** Define a function first, then pass it to map, filter or reduce (see the cells below).

```python
# Task 3: define a function first, then use it inside map
def square(x):
    return x * x

numbers = [1, 2, 3, 4]
squared_numbers = map(square, numbers)
print(list(squared_numbers))
```

```python
# Task 3 (filter): define a function that returns True or False first
def is_even(x):
    return x % 2 == 0

print(list(filter(is_even, numbers)))
```

```python
# Task 3 (reduce): define a function with two inputs first
from functools import reduce

def add(x, y):
    return x + y

print(reduce(add, numbers))
```

### Exercises: Level 2 - Answers

```python
# Use map to change each country to uppercase
countries_upper = list(map(str.upper, countries))
print(countries_upper)
```

```python
# Use map to square each number
squared_numbers = list(map(lambda x: x**2, numbers))
print(squared_numbers)
```

```python
# Use map to change each name to uppercase
uppercase_names = list(map(str.upper, names))
print(uppercase_names)
```

```python
# For the filter tasks I use a list of north European countries
countries = ["Estonia", "Finland", "Sweden", "Denmark", "Norway", "Iceland"]

# Filter: countries containing 'land'
land_countries = list(filter(lambda country: 'land' in country, countries))
print(land_countries)
```

```python
# Filter: countries having exactly six characters
six_char_countries = list(filter(lambda country: len(country) == 6, countries))
print(six_char_countries)
```

```python
# Filter: countries with six letters or more
six_or_more = list(filter(lambda country: len(country) >= 6, countries))
print(six_or_more)
```

```python
# Filter: countries starting with 'E'
e_countries = list(filter(lambda country: country.startswith('E'), countries))
print(e_countries)
```

```python
# Chain map and filter: keep countries with 'land', then make them uppercase
result = list(map(str.upper, filter(lambda country: 'land' in country, countries)))
print(result)
```

```python
# Chain map, filter and reduce: square the numbers, keep the even ones, then add them
from functools import reduce

total = reduce(lambda x, y: x + y,
               filter(lambda x: x % 2 == 0,
                      map(lambda x: x ** 2, numbers)))
print(total)
```

```python
# Function that returns only the string items of a list
def get_string_lists(lst):
    return [item for item in lst if isinstance(item, str)]

print(get_string_lists([1, 'apple', 3.5, 'banana', True, 'cat']))
```

```python
# Use reduce to sum all the numbers
from functools import reduce
total = reduce(lambda x, y: x + y, numbers)
print(total)
```

```python
# Use reduce to join the countries into one sentence
# Target: Estonia, Finland, Sweden, Denmark, Norway, and Iceland are north European countries
joined = reduce(lambda x, y: f"{x}, {y}", countries[:-1])
result_sentence = f"{joined}, and {countries[-1]} are north European countries"
print(result_sentence)
```

For the next tasks I need the full countries list. I load `countries-data.py` (from the 30-Days-Of-Python repository). The code first looks for the file next to this notebook and downloads it if it is not there.

```python
import ast
import os
import urllib.request

# Load the countries data (a list of dictionaries)
file_name = 'countries-data.py'
url = 'https://raw.githubusercontent.com/Asabeneh/30-Days-Of-Python/master/data/countries-data.py'

if os.path.exists(file_name):
    text = open(file_name, encoding='utf-8').read()
else:
    text = urllib.request.urlopen(url).read().decode('utf-8')

countries_data = ast.literal_eval(text.strip())
all_countries = [country['name'] for country in countries_data]

print(len(all_countries), 'countries loaded')
print(all_countries[:3])
```

```python
# Function that returns countries with a common pattern
def categorize_countries(pattern):
    return [country for country in all_countries if pattern in country]

print(categorize_countries('land'))
print(categorize_countries('stan'))
```

```python
# Function returning a dictionary: starting letter -> number of countries
def country_starting_letters():
    letter_dict = {}
    for country in all_countries:
        first_letter = country[0]
        if first_letter in letter_dict:
            letter_dict[first_letter] += 1
        else:
            letter_dict[first_letter] = 1
    return letter_dict

print(country_starting_letters())
```

```python
# First ten countries
def get_first_ten_countries():
    return all_countries[:10]

print(get_first_ten_countries())
```

```python
# Last ten countries
def get_last_ten_countries():
    return all_countries[-10:]

print(get_last_ten_countries())
```

### Exercises: Level 3 - Answers

```python
# Sort countries by name
sorted_by_name = sorted(countries_data, key=lambda x: x['name'])
print([c['name'] for c in sorted_by_name[:5]])
```

```python
# Sort countries by capital
sorted_by_capital = sorted(countries_data, key=lambda x: x['capital'])
print([(c['capital'], c['name']) for c in sorted_by_capital[:5]])
```

```python
# Sort countries by population (smallest first)
sorted_by_population = sorted(countries_data, key=lambda x: x['population'])
print([(c['name'], c['population']) for c in sorted_by_population[:5]])
```

```python
# Ten most spoken languages (by how many countries speak them)
from collections import Counter

languages = [lang for country in countries_data for lang in country['languages']]
most_common_languages = Counter(languages).most_common(10)
for language, count in most_common_languages:
    print(language, '-', count)
```

```python
# Ten most populated countries
most_populated_countries = sorted(countries_data, key=lambda x: x['population'], reverse=True)[:10]
for country in most_populated_countries:
    print(country['name'], '-', country['population'])
```
