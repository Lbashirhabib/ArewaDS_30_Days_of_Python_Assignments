## Creating a package
A package is a folder with an `__init__.py` file and some modules.

```python
import os
os.makedirs('mypackage', exist_ok=True)
```

```python
%%writefile mypackage/arithmetics.py
def add_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total

def subtract(a, b):
    return (a - b)

def multiple(a, b):
    return a * b

def division(a, b):
    return a / b

def remainder(a, b):
    return a % b

def power(a, b):
    return a ** b
```

```python
%%writefile mypackage/greet.py
def greet_person(firstname, lastname):
    return f'{firstname} {lastname}, welcome to 30DaysOfPython Challenge!'
```

```python
from mypackage import arithmetics, greet

print(arithmetics.add_numbers(1, 2, 3, 5))
print(arithmetics.subtract(5, 3))
print(arithmetics.multiple(5, 3))
print(arithmetics.division(5, 3))
print(arithmetics.remainder(5, 3))
print(arithmetics.power(5, 3))
print(greet.greet_person('Asabeneh', 'Yetayeh'))
```

### 1. Read Romeo and Juliet and find the 10 most frequent words

```python
import re
from collections import Counter

romeo_and_juliet = 'https://www.gutenberg.org/files/1112/1112.txt'
response = requests.get(romeo_and_juliet)
text = response.text

# lower case, keep only words (letters and apostrophes)
words = re.findall(r"[a-z']+", text.lower())
print('Total words:', len(words))

word_counts = Counter(words)
for word, count in word_counts.most_common(10):
    print(word, count)
```

### 2. Cats API

```python
import pandas as pd
import numpy as np

cats_api = 'https://api.thecatapi.com/v1/breeds'
cats = requests.get(cats_api).json()
print(len(cats), 'breeds')
if isinstance(cats, list):
    print(cats[0]['name'], cats[0]['weight'], cats[0]['life_span'])
else:
    print("The Cat API request failed:", cats.get('message', cats))
    cats = []
```

Weight and life span come as text like `'3 - 5'`, so I take the middle of the range for each breed.

```python
import re
import numpy as np
import pandas as pd


def range_to_number(text):
    """Convert a range string to its midpoint.

    Examples:
        '3 - 5' -> 4.0
        '7'     -> 7.0
        '' / None / no digits -> np.nan
    """
    if text is None:
        return np.nan
    parts = [float(x) for x in re.findall(r'\d+\.?\d*', str(text))]
    return sum(parts) / len(parts) if parts else np.nan


def _safe_get(d, *keys):
    """Walk nested dicts safely, returning None on any miss."""
    for k in keys:
        if not isinstance(d, dict):
            return None
        d = d.get(k)
        if d is None:
            return None
    return d


weights   = pd.Series([range_to_number(_safe_get(c, 'weight', 'metric')) for c in cats])
lifespans = pd.Series([range_to_number(_safe_get(c, 'life_span'))         for c in cats])
```

#### a) min, max, mean, median and standard deviation of weight (kg)

```python
print('min:   ', weights.min())
print('max:   ', weights.max())
print('mean:  ', weights.mean())
print('median:', weights.median())
print('std:   ', weights.std())
```

#### b) the same for lifespan (years)

```python
print('min:   ', lifespans.min())
print('max:   ', lifespans.max())
print('mean:  ', lifespans.mean())
print('median:', lifespans.median())
print('std:   ', lifespans.std())
```

### 4. UCI datasets page with BeautifulSoup

```python
from bs4 import BeautifulSoup

uci_url = 'https://archive.ics.uci.edu/ml/datasets.php'
response = requests.get(uci_url)
print(response.status_code)

soup = BeautifulSoup(response.content, 'html.parser')
print(soup.title.text)

# print the first 20 links on the page
links = soup.find_all('a')
print('Number of links:', len(links))
for link in links[:20]:
    print(link.text.strip(), '->', link.get('href'))
```
