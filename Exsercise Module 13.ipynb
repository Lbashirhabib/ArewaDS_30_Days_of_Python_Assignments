````python
import datetime
print(dir(datetime))
````

````python
# 1. current day, month, year, hour, minute and timestamp
from datetime import datetime
now = datetime.now()
print(now)
print('Day:', now.day)
print('Month:', now.month)
print('Year:', now.year)
print('Hour:', now.hour)
print('Minute:', now.minute)
print('Timestamp:', now.timestamp())
````

````python
# 2. format the current date
time_one = now.strftime("%m/%d/%Y, %H:%M:%S")
print(time_one)
````

````python
# 3. change the string "5 December, 2019" to a date
date_string = "5 December, 2019"
print("date_string =", date_string)
date_object = datetime.strptime(date_string, "%d %B, %Y")
print("date_object =", date_object)
````

````python
# 4. time difference between now and new year (1 January next year)
new_year = datetime(now.year + 1, 1, 1)
the_diff = new_year - now
print('Time left until new year =', the_diff)
print('Days left:', the_diff.days)
````

````python
# 5. time difference between 1 January 1970 and now
epoch = datetime(1970, 1, 1)
diff = datetime.now() - epoch
print('Time since 1 January 1970 =', diff)
print('Days since 1970:', diff.days)
````

### File handling answers

The data files are in the `data` folder, which is next to this notebook folder, so I use `../data/`.

````python
import json
import re
import sys
from collections import Counter

DATA = '../data/'
````

**Level 1, question 1: count lines and words of a text file**

````python
def count_lines_and_words(filename):
    with open(filename, encoding='utf-8') as f:
        text = f.read()
    number_of_lines = len(text.splitlines())
    number_of_words = len(text.split())
    return number_of_lines, number_of_words

files = ['obama_speech.txt', 'michelle_obama_speech.txt',
         'donald_speech.txt', 'melina_trump_speech.txt']

for file in files:
    lines, words = count_lines_and_words(DATA + file)
    print(file, '-> lines:', lines, 'words:', words)
````

**Question 2: ten most spoken languages**

````python
def most_spoken_languages(filename, number):
    with open(filename, encoding='utf-8') as f:
        countries = json.load(f)
    languages = []
    for country in countries:
        for language in country['languages']:
            languages.append(language)
    counts = Counter(languages).most_common(number)
    # change to (count, language) like the example
    return [(count, language) for language, count in counts]

print(most_spoken_languages(DATA + 'countries_data.json', 10))
print(most_spoken_languages(DATA + 'countries_data.json', 3))
````

**Question 3: ten most populated countries**

````python
def most_populated_countries(filename, number):
    with open(filename, encoding='utf-8') as f:
        countries = json.load(f)
    sorted_countries = sorted(countries, key=lambda c: c['population'], reverse=True)
    result = []
    for country in sorted_countries[:number]:
        result.append({'country': country['name'], 'population': country['population']})
    return result

print(most_populated_countries(DATA + 'countries_data.json', 10))
print(most_populated_countries(DATA + 'countries_data.json', 3))
````

**Question 5: find_most_common_words**

````python
import os

def find_most_common_words(text, number):
    # text can be a file name or a normal string
    if os.path.isfile(text):
        with open(text, encoding='utf-8', errors='ignore') as f:
            text = f.read()
    words = re.findall(r"[a-zA-Z']+", text)
    counts = Counter(words).most_common(number)
    return [(count, word) for word, count in counts]

sample = 'the cat and the dog and the bird went to the park'
print(find_most_common_words(sample, 3))
````

**Question 7: check similarity between two texts**

````python
sys.path.append(DATA)
from stop_words import stop_words

def clean_text(text):
    # lower case, keep only letters and spaces
    text = text.lower()
    text = re.sub(r"[^a-z\s]", ' ', text)
    return text

def remove_support_words(text):
    # remove the stop words
    words = text.split()
    return [word for word in words if word not in stop_words]

def check_text_similarity(text1, text2):
    # text1 and text2 can be file names or strings
    texts = []
    for t in (text1, text2):
        if os.path.isfile(t):
            with open(t, encoding='utf-8', errors='ignore') as f:
                t = f.read()
        texts.append(set(remove_support_words(clean_text(t))))
    common = texts[0] & texts[1]
    all_words = texts[0] | texts[1]
    return round(len(common) / len(all_words) * 100, 2)

similarity = check_text_similarity(DATA + 'michelle_obama_speech.txt', DATA + 'melina_trump_speech.txt')
print('Similarity between Michelle and Melania speeches:', similarity, '%')
````

````python
python_lines = 0
javascript_lines = 0
java_lines = 0

with open(DATA + 'hacker_news.csv', encoding='utf-8', errors='ignore') as f:
    for line in f:
        if 'python' in line or 'Python' in line:
            python_lines += 1
        if 'JavaScript' in line or 'javascript' in line or 'Javascript' in line:
            javascript_lines += 1
        if 'Java' in line and 'JavaScript' not in line:
            java_lines += 1

print('Lines with python or Python:', python_lines)
print('Lines with JavaScript, javascript or Javascript:', javascript_lines)
print('Lines with Java and not JavaScript:', java_lines)
````
