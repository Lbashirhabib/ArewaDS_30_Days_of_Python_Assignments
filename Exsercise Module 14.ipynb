## Exercises: Level 1

### 1. Function that generates a six character random_user_id

````python
# the examples above imported the name random as a function, so I import the module again
import random
import string

def random_user_id():
    characters = string.ascii_letters + string.digits
    user_id = ''
    for i in range(6):
        user_id += random.choice(characters)
    return user_id

print(random_user_id())
````

### 2. user_id_gen_by_user (takes the number of characters and the number of ids from input())

````python
def user_id_gen_by_user():
    # the user types two numbers like: 5 5
    number_of_characters, number_of_ids = input('Number of characters and number of ids: ').split()
    number_of_characters = int(number_of_characters)
    number_of_ids = int(number_of_ids)
    characters = string.ascii_letters + string.digits
    ids = []
    for i in range(number_of_ids):
        user_id = ''
        for j in range(number_of_characters):
            user_id += random.choice(characters)
        ids.append(user_id)
    return '\n'.join(ids)

print(user_id_gen_by_user())   # try: 5 5
````

### 3. rgb_color_gen

````python
def rgb_color_gen():
    r = random.randint(0, 255)
    g = random.randint(0, 255)
    b = random.randint(0, 255)
    return f'rgb({r},{g},{b})'

print(rgb_color_gen())
````

## Exercises: Level 2

### 1. list_of_hexa_colors

````python
def list_of_hexa_colors(n=1):
    hexa_symbols = '0123456789abcdef'
    colors = []
    for i in range(n):
        color = '#'
        for j in range(6):
            color += random.choice(hexa_symbols)
        colors.append(color)
    return colors

print(list_of_hexa_colors(3))
````

### 2. list_of_rgb_colors

````python
def list_of_rgb_colors(n=1):
    colors = []
    for i in range(n):
        colors.append(rgb_color_gen())
    return colors

print(list_of_rgb_colors(3))
````

### 3. generate_colors

````python
def generate_colors(color_type, n):
    if color_type == 'hexa':
        return list_of_hexa_colors(n)
    elif color_type == 'rgb':
        return list_of_rgb_colors(n)
    else:
        return 'color_type must be hexa or rgb'

print(generate_colors('hexa', 3))
print(generate_colors('hexa', 1))
print(generate_colors('rgb', 3))
print(generate_colors('rgb', 1))
````

## Exercises: Level 3

### 1. shuffle_list

````python
def shuffle_list(lst):
    shuffled = lst.copy()      # copy so the original list is not changed
    random.shuffle(shuffled)
    return shuffled

numbers = [1, 2, 3, 4, 5, 6]
print(shuffle_list(numbers))
print(numbers)
````

### 2. Seven unique random numbers between 0 and 9

````python
def seven_unique_numbers():
    return random.sample(range(0, 10), 7)   # sample never repeats a number

print(seven_unique_numbers())
````
